```python
from langchain.agents import create_agent
def get_weather(city: str)->str:
    '''get the weather'''
    return f'The weather for the {city} is sunny'
### create an agents
agent=create_agent(
    model='gpt-5',
    tools=[get_weather],
    system_prompt='you are helful agents',
    # verbose

)
agent
```

```python
#### run the agents 
respponse=agent.invoke({"message":[{'role':'user',"context":"What is weather like in New York"}]})
```
