### Streaming and Batch 

Streaming in AI means receiving the model's response piece by piece instead of waiting for the entire response to finish. It's how ChatGPT can display an answer while the AI is still generating it.

```python
model_gemini.invoke('write me a 200 words paragraph on AI')
```

```python
for chhunk in  model_gemini.stream('write me 200 words on myself'):
    print(chhunk.text)
```

```python
for chhunk in model_gemini.stream('write me 200 words on myself'):
    print(chhunk.text,end='|',flush=true)
```
